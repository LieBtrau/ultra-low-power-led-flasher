# Output timing calculator
$$\Delta t = \frac{L * \Delta I}{U}$$
The output current is linear with the ON-time of the switch and the inductor value.
* The inductor value is already quite large, so we will not touch on that one.
* The ON-time can be increased.  Be careful not to increase it too much, to avoid inductor core saturation.  

$$ t_{W} = K * R_{EXT} * C_{EXT} $$
* $t_{W}$ = typical output pulse width in ns;
* $R_{EXT}$ = external resistor in kΩ; typically < 200 kOhm
* $C_{EXT}$ = external capacitor in pF; typically > 10 pF
* $K$ = unitless correction factor, around 1.4 for 1.8 V supply voltage

To minimize power consumption, the largest possible resistor, i.e. 200 kOhm trimpot will be used.
- For a maximum $t_{W}$ of 20 µs, we'll need about 100 pF of capacitance.
- Minimum $t_{W}$ of 2 µs with a 100 pF capacitor will lead to a 20 kOhm resistor.  This resistor will be placed in series with the trimpot.

As such, the total resistance at maximum pulse width will actually be 220 kOhm.

```python
K = 1.4 # in the range of 1 to 1.4.  Depends on supply voltage.
R_EXT = 200 # [kOhm]
C_EXT = 100 # [pF]
t_W = K * R_EXT * C_EXT
print(f"Pulse width: {t_W/1e3:.1f} µs.")
```
